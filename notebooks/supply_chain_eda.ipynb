```python
import pandas as pd
```

```python
# 1. Load the dataset (DataCo often requires 'latin1' or 'ISO-8859-1' encoding due to special characters)
print("Loading dataset...")
df = pd.read_csv('DataCoSupplyChainDataset.csv', encoding='latin1')
```

```python
# 2. Initial Inspection
print("\n--- Dataset Shape ---")
print(df.shape)
```

```python
print("\n--- Missing Values Summary ---")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
# 3. Convert Date Columns to Datetime Objects
# Note: Check your exact column names if they differ slightly
date_columns = ['order date (Date)', 'shipping date (Date)']
for col in date_columns:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors='coerce')
```

```python
# 4. Feature Engineering: Calculate Shipping Duration & Delay Status
if 'order date (Date)' in df.columns and 'shipping date (Date)' in df.columns:
    # Actual shipping days (Difference between shipping date and order date)
    df['Shipping_Days'] = (df['shipping date (Date)'] - df['order date (Date)']).dt.days
```

```python
# 5. Create a Delay Flag based on Delivery Status
# 'Late delivery', 'Advance shipping', 'Shipping on time', 'Shipping canceled'
if 'Delivery Status' in df.columns:
    df['Is_Delayed'] = df['Delivery Status'].apply(lambda x: 1 if x == 'Late delivery' else 0)
```

```python
print("\n--- Cleaned Data Sample with New Features ---")
```

```python
import pandas as pd

# Ensure column names are clean of accidental whitespace
df.columns = df.columns.str.strip()
```

```python
# 1. Overall Delivery Status Breakdown (%)
print("--- Delivery Status Breakdown ---")
if 'Delivery Status' in df.columns:
    print(df['Delivery Status'].value_counts(normalize=True) * 100)

# 2. Delay Rate by Shipping Mode
# (DataCo usually has 'Shipping Mode' and 'Late_delivery_risk' (1 or 0))
print("\n--- Late Delivery Risk by Shipping Mode (%) ---")
if 'Shipping Mode' in df.columns and 'Late_delivery_risk' in df.columns:
    print(df.groupby('Shipping Mode')['Late_delivery_risk'].mean() * 100)
```

```python
# 3. Delay Risk by Market / Region (if available)
if 'Market' in df.columns and 'Late_delivery_risk' in df.columns:
    print("\n--- Late Delivery Risk by Market (%) ---")
    print(df.groupby('Market')['Late_delivery_risk'].mean() * 100)
```

```python
# Export the cleaned dataset to a new CSV file
df.to_csv('cleaned_supply_chain_data.csv', index=False)
print("Cleaned data successfully saved as 'cleaned_supply_chain_data.csv'!")
```
